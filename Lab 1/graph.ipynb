```python

import matplotlib.pyplot as plt

threads = [1, 2, 4, 8, 16]

speedup_100k = [2.508, 6.254, 13.741, 24.236, 0.022]
speedup_1m = [3.703, 12.329, 17.866, 13.078, 0.283]
speedup_10m = [6.631, 9.246, 7.163, 5.680, 2.994]
speedup_100m = [4.640, 7.185, 9.586, 9.488, 6.740]

plt.figure(figsize=(10, 6))

plt.plot(threads, speedup_100k, marker='o', label='N = 10^5')
plt.plot(threads, speedup_1m, marker='s', label='N = 10^6')
plt.plot(threads, speedup_10m, marker='^', label='N = 10^7')
plt.plot(threads, speedup_100m, marker='d', label='N = 10^8')

plt.xlabel('Number of Threads')
plt.ylabel('Speedup (Serial Time / Parallel Time)')
plt.title('OpenMP Vector Addition: Speedup vs. Thread Count')

plt.xticks(threads)
plt.axhline(y=1, linestyle='--', label='No speedup')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()

plt.savefig('task1_speedup.png', dpi=300)
plt.show()
```
